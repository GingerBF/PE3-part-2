```python
import numpy as np
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
frequencies = np.logspace(3, 10)
RG = 10
R1 = 10000
R2 = 60
C = 150e-12

Z1 = R1
Z2 = R2 + 1/(1j * frequencies * C * 2 * np.pi)
ZF = (Z1 * Z2)/(Z1 + Z2)
ZG = RG

Gain = 1 + np.abs(ZF/ZG)

plt.figure(figsize=(8,5))
plt.semilogx(frequencies, np.abs(Gain), label='theoretical gain')
plt.axvline(10000, color='red', linestyle='--', label='10 kHz')
plt.axhline(7, color='green', linestyle='--', label='7x gain')
plt.xlabel('Frequency [Hz]')
plt.ylabel('|Gain|')
plt.title('Gain vs Frequency')
plt.grid(True, which='both', alpha=0.3)
plt.tight_layout()
plt.legend()
plt.show()
```
